## 5. 測定誤差のある線形モデル
### 問題
#### 問題5.7

```python
m, s, a, b = sp.symbols("\\mu, \\sigma, \\alpha, \\beta", real=True)
x, y, sxx, syy, sxy = sp.symbols("\\bar{x}, \\bar{y}, S_{xx}, S_{yy}, S_{xy}", real=True)
sx, sy, la = sp.symbols("\\sigma_x, \\sigma_y, \\lambda", real=True)
```

```python
A = 1 / s**2 + 1 / sx**2 + b**2 / sy**2
Bsq = m**2 / s**4 + (sxx + x**2) / sx**4 + b**2 / sy**4 * (syy + (y-a)**2) + 2 * m * x / (s**2 * sx**2) + 2 * m * b * (y-a) / (s**2 * sy**2) + 2 * b * (sxy + x * y - a * x) / (sx**2 * sy**2)
C = m**2 / (2 * s**2) + (sxx + x**2) / (2 * sx**2) + (syy + (y-a)**2) / (2 * sy**2)
```

```python
l = - sp.log(s) - sp.log(A) / 2 + Bsq / (2*A) - C
l = l.subs(sy, la*sx)
```

```python
dldm = sp.diff(l, m).factor()
numm, denm = dldm.as_numer_denom()

dlds = sp.diff(l, s).factor()
nums, dens = dlds.as_numer_denom()

dlda = sp.diff(l, a).factor()
numa, dena = dlda.as_numer_denom()

dldb = sp.diff(l, b).factor()
numb, denb = dldb.as_numer_denom()
```

```python
masol = sp.solve([numm, numa], [m, a])
msol = masol[m]
asol = masol[a]

display(msol)
display(asol)
```

```python
nums = nums.subs({m: msol, a: asol}).factor()
numb = numb.subs({m: msol, a: asol}).factor()

display(nums)
display(numb)
```

```python
nums = nums / s
numb = numb / s**2
```

```python
sbsol = sp.solve([nums, numb], [s, b]) # eight solutions
```
