### 練習問題
#### 問題1.4

```r
N <- 11

p <- matrix(0, nrow = N, ncol = N)
```

$p(1, n) = 1 / n$および$p(r, 1) = 0　~ (r \geq 0)$

```r
for (n in 1:N) {
  p[1, n] <- 1 / n
}
```

```r
for (n in 2:N){
  for (r in 2:N){
    p[r, n] <- p[r-1, n-1] / n + p[r, n-1]*(n-1) / n
  }
}
```

```r
for (r in 1:N){
  cat(r, p[r, N], "\n")
}
```

```r
par(bg = "white")
barplot(p[, N], xlab = "r", ylab = "p(r, N)")
```
