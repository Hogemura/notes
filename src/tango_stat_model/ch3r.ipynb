## 3. bootstrap
### 3.3 bootstrapによる推測
#### 表3.2

```r
bootstrap_med <- function(nb){
  g <- rep(0, nb) #nb is a value of B
  x <- c(16, 41, 57, 76, 124, 129, 193, 215, 280, 363, 914)
  med <- x[6]
  for (i in 1:nb) {y <- sort(sample(x, replace=T)); g[i] <- y[6]}
  gmean <- mean(g); gsd <- sqrt(var(g))
  gbias <- gmean - med; gmod <- med - gbias

  return(c(gmean, gsd, gbias, gmod))
}

data.frame(bootstrap_med(100), bootstrap_med(200), bootstrap_med(500), bootstrap_med(1000), bootstrap_med(2000))
```

#### 表3.3

```r
bootstrap_mean <- function(nb){
  g <- rep(0, nb) #nb is a value of B
  x <- c(16, 41, 57, 76, 124, 129, 193, 215, 280, 363, 914)
  med <- x[6] # \tilde\theta
  mean <- mean(x) # \hat\theta
  for (i in 1:nb) {y <- sort(sample(x, replace=T)); g[i] <- mean(y)} # \hat\theta^*
  gmean <- mean(g); gsd <- sqrt(var(g))
  gbias <- gmean - med; gmod <- mean - gbias

  return(c(gmean, gsd, gbias, gmod))
}

data.frame(bootstrap_mean(100), bootstrap_mean(200), bootstrap_mean(500), bootstrap_mean(1000), bootstrap_mean(2000))
```
